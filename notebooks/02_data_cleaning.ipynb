## Handling Missing Values

Missing values were investigated before cleaning. Columns such as `promotion-ids`,
`fulfilled-by`, and `Courier Status` contain missing values that can represent
non-applicable or unavailable information rather than data-entry errors.

For shipping location fields, rows with missing values will be removed because
these fields are important for regional analysis.

```python
# Remove rows with missing shipping location information
df = df.dropna(
    subset=["ship-city", "ship-state", "ship-postal-code", "ship-country"]
)

# Fill non-applicable categorical fields
df["Courier Status"] = df["Courier Status"].fillna("Not Available")
df["promotion-ids"] = df["promotion-ids"].fillna("No Promotion")
df["fulfilled-by"] = df["fulfilled-by"].fillna("Not Specified")

# Check remaining missing values
df.isnull().sum()
```
